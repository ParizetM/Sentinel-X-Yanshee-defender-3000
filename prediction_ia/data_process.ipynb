```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import IsolationForest
```

```python
df = pd.read_csv("telemetry.csv")
```

```python
df.head(20)
```

```python
df.describe()
```

```python
# Features de télémétrie seules : on retire les colonnes non pertinentes (métadonnées / état)
COLONNES_EXCLUES = ["actuators", "device", "presence", "timestamp", "topic", "uptime_s"]

df_filtre = df.drop(columns=COLONNES_EXCLUES)

df_filtre.head()
```

```python
df_filtre['gas'] = df_filtre['gas'].str.extract(r"'raw':\s*(\d+)")
```

```python
df_filtre['gas'] = df_filtre['gas'].astype(int)
```

```python
df_filtre
```

```python
# Sauvegarde du dataset filtré final (features télémétrie brutes, gas extrait en int)
CHEMIN_SORTIE = "telemetry_filtre.csv"

df_filtre.to_csv(CHEMIN_SORTIE, index=False)

print(f"df_filtre sauvegardé -> {CHEMIN_SORTIE} ({df_filtre.shape[0]} lignes x {df_filtre.shape[1]} colonnes)")
```
